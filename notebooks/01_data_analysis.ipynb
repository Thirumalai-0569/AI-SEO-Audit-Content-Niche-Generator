# 01 — Data Analysis
Explore the sample post dataset. **It is synthetic demo data**, not actual platform analytics.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
DATA = Path('../data/sample_posts.csv')
df = pd.read_csv(DATA)
df.head()
```

```python
df.info()
```

```python
df.describe(include='all').T
```

```python
df.groupby('platform').agg(posts=('post_id','count'), mean_views=('views','mean'), mean_engagement=('historical_engagement_rate','mean'), success_rate=('success','mean')).round(2)
```

```python
df.groupby('topic')['views'].mean().sort_values(ascending=False).plot(kind='bar', figsize=(12,4), title='Average synthetic views by topic')
plt.tight_layout()
```
